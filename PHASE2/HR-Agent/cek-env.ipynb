```python
import os
from pathlib import Path

from dotenv import load_dotenv


def find_env_file() -> Path | None:
    """Cari .env dari folder notebook hingga folder induknya."""
    current = Path.cwd().resolve()

    for folder in (current, *current.parents):
        env_file = folder / ".env"
        if env_file.exists():
            return env_file

    return None


def is_sensitive(name: str) -> bool:
    sensitive_words = ("KEY", "TOKEN", "SECRET", "PASSWORD", "PASS", "CREDENTIAL")
    return any(word in name.upper() for word in sensitive_words)


def display_value(name: str, value: str) -> str:
    if is_sensitive(name):
        return "(terdaftar - nilai disamarkan)"
    return value if value else "(kosong)"


env_file = find_env_file()

if env_file:
    load_dotenv(env_file)
    print(f"File .env ditemukan: {env_file}\n")
else:
    print("File .env tidak ditemukan. Mengecek environment variable dari sistem/terminal.\n")

# Sesuaikan prefix ini dengan konfigurasi project Anda bila diperlukan.
prefixes = (
    "OPENAI_",
    "ANTHROPIC_",
    "GOOGLE_",
    "GEMINI_",
    "API_",
    "APP_",
    "DATABASE_",
    "REDIS_",
    "MODEL_",
    "DEBUG",
    "ENV",
)

configurations = {
    key: value
    for key, value in os.environ.items()
    if key.startswith(prefixes)
}

if not configurations:
    print("Tidak ada konfigurasi project yang terdeteksi.")
else:
    print("Konfigurasi yang terdaftar:")
    for key in sorted(configurations):
        print(f"- {key} = {display_value(key, configurations[key])}")
```
